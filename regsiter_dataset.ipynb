```python
# Handle to the workspace
from azure.ai.ml import MLClient

# Authentication package
from azure.identity import DefaultAzureCredential

import mltable

from azure.ai.ml.entities import Data
from azure.ai.ml.constants import AssetTypes
import json
import os
```

```python
credential = DefaultAzureCredential()

ml_client = MLClient(
    credential,
    subscription_id="08539d0e-620d-424c-aea8-56853ed8fe3e",
    resource_group_name="dspro2",
    workspace_name="dspro2ml",
)

tbl = mltable.from_delta_lake(
    delta_table_uri="abfss://gold@dspro2.dfs.core.windows.net/bostonhousing/"
) 

base_save_path = f"azureml://subscriptions/08539d0e-620d-424c-aea8-56853ed8fe3e/resourcegroups/dspro2/workspaces/dspro2ml/datastores/deltalake_gold"
ml_table_path = f"{base_save_path}/paths/stats"


tbl.save(
    path= ml_table_path,
    colocated=False,
    show_progress=True,
    overwrite=True
)

my_data = Data(
    path=ml_table_path,
    type=AssetTypes.MLTABLE,
    name="boston_gold"
)

ml_client.data.create_or_update(my_data)
```
